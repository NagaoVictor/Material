```python
import matplotlib.pyplot as plt
from sympy import *
import mplcyberpunk
import numpy as np
```

```python
x = Symbol('x')
y = Symbol('y')
x1 = int(input('Digite a coordenada x: '))
y1 = int(input('Digite a coordenada y: '))
x2 = int(input('Digite a coordenada x1: '))
y2 = int(input('Digite a coordenada y1: '))
scalex = [0, x2+2]
scaley = [0, y2+2]
```

```python
M = Matrix([[x,y,1],[x1,y1,1],[x2,y2,1]])
Mdet = M.det()
print(f'{M} {Mdet}')
```

```python
M.det().subs(y,0)
```

PARA FUNÇÕES COM A VARIÁVEL Y DEPENDENTE, TEMOS O CÓDIGO
solveset(func, y)
PARA FUNÇÕES COM A VARIÁVEL X DEPENDENTE, TEMOS O CÓDIGO
solveset(func, x)

```python
solveset(M.det(), y)
```

```python
solveset(M.det(), x)
```

```python
solveset(M.det(), y).subs(x,0)
```

```python
solveset(M.det(), x).subs(y,0)
```

```python
plt.figure(facecolor='silver')
ax = plt.axes()
ax.set_facecolor('#2e2157')
plt.title('Equação da reta')
plt.plot(scalex, scaley, color='cyan')
plt.ylabel(f'f(x) = {solveset(M.det(), y)}')
plt.xlabel(x)
mplcyberpunk.add_glow_effects()
plt.show()
```

```python
# Neon Style!!

x = Symbol('x')
y = Symbol('y')
x0 = Symbol('x0')
y0 = Symbol('y0')
a = Symbol('a')
b = Symbol('b')
```

```python
elipse = (x - x0)**2/a**2 + (y - y0)**2/b**2 - 1
elipse
```

```python
genericElispe = elipse.subs(x0 , 0).subs(y0, 0).subs(b, 2).subs(a, 8)
```

```python
genericElispe
```

```python
p, q = solveset(genericElispe, y)
p1,q1 = solveset(genericElispe, x)
```

```python
listElipsepAxis4 = []
listElipsepAxis1 = []
for i in np.linspace(0,8,100):
    prin = p.subs(x, i)
    if prin > -0.56 and prin < -0.43:
        achei = len(listElipsepAxis4)
    if prin > -1.01 and prin < -0.99:
        achei1 = len(listElipsepAxis4)
    if prin > -1.51 and prin < -1.49:
        achei2 = len(listElipsepAxis4)
    listElipsepAxis4.append(prin)
    listElipsepAxis1.append(-prin)
```

```python
aixox = [0, achei, achei]
aixoy = [0, -0.5, 0]
aixox1 = [0, achei1, achei1]
aixoy1 = [0, -1.0, 0]
aixox2 = [0, achei2, achei2]
aixoy2 = [0, -1.5, 0]
fig = plt.figure(facecolor='silver', figsize=(10,6))
ax = plt.axes()
plt.plot(range(len(listElipsepAxis4)), listElipsepAxis4, color = 'cyan')
mplcyberpunk.add_gradient_fill(gradient_start='max')
plt.plot(range(len(listElipsepAxis1)), listElipsepAxis1, color = 'cyan')
ax.set_facecolor('#2e2157')
plt.plot(aixox, aixoy, color='lime')
plt.plot(aixox1, aixoy1, color='fuchsia')
plt.plot(aixox2, aixoy2, color='gold')
mplcyberpunk.add_gradient_fill(gradient_start='max')
plt.ylabel(f'{elipse+1} = 1')
plt.xlabel(x)
plt.title('Elípse')
plt.show()
```
